# Lab 2: Data Analysis and Visualization
**Dataset:** Adult Income (1994 US Census)

This notebook works through visual encoding, chart selection, and outlier detection using the Adult Income dataset.

## Setup and Cleaning

The dataset marks missing values with the string `"?"` instead of a real NaN, in three columns: `workclass`, `occupation`, `native.country`. We convert those to proper NaN and drop the rows, since they're a small fraction of the data (~5-6%) and several of our tasks group by these categorical columns — leaving `"?"` in would create a fake "unknown" category that distorts every bar chart average.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

sns.set_style("whitegrid")

df = pd.read_csv("adult.csv")
df.replace("?", np.nan, inplace=True)
print("Before dropping missing:", df.shape)
df.dropna(inplace=True)
df.reset_index(drop=True, inplace=True)
print("After dropping missing:", df.shape)
df.head()
```

## Task 1: Bar Chart — Compare a Category

**Question:** Does average `hours.per.week` differ across `occupation`?

Bar charts use **length + position**, the two most accurately-judged visual channels — this is why we use a bar chart instead of, say, bubble sizes to compare these averages.

```python
avg_hours_by_occ = df.groupby("occupation")["hours.per.week"].mean().sort_values(ascending=True)

plt.figure(figsize=(10, 6))
avg_hours_by_occ.plot(kind="bar", color="#4C72B0", edgecolor="black")
plt.title("Average Hours Worked per Week by Occupation")
plt.xlabel("Occupation")
plt.ylabel("Average Hours per Week")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig("task1_bar.png", dpi=100)
plt.show()

print(avg_hours_by_occ)
```

## Task 2: Count Plot — Category Frequency

**Question:** How many records fall into each `education` level?

Unlike Task 1 (which needed a groupby + mean), a count plot just tallies rows per category — seaborn's `countplot` does that counting internally.

```python
order = df["education"].value_counts().index

plt.figure(figsize=(10, 6))
sns.countplot(data=df, x="education", order=order, color="#4C72B0", edgecolor="black")
plt.title("Number of Records by Education Level")
plt.xlabel("Education")
plt.ylabel("Count")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig("task2_countplot.png", dpi=100)
plt.show()

df["education"].value_counts()
```

## Task 3: Histogram — Distribution Shape

**Question:** What does the distribution of `age` look like — symmetric, skewed left, or skewed right?

```python
plt.figure(figsize=(10, 6))
plt.hist(df["age"], bins=30, color="#4C72B0", edgecolor="black")
plt.title("Distribution of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig("task3_histogram.png", dpi=100)
plt.show()

print("Mean age:", df["age"].mean().round(2))
print("Median age:", df["age"].median())
print("Skew:", df["age"].skew().round(3))
```

## Task 4: Box Plot — Visual Outlier Detection

**Why a box plot:** this is the chart that actually draws the IQR calculation from your manual. The box shows the middle 50% of the data (Q1 to Q3), the whiskers extend to 1.5×IQR beyond the box, and anything past that gets plotted as an individual dot — a visible outlier. We're using the same `age` column from Task 3, split by `sex` to see if the spread differs between groups.

```python
plt.figure(figsize=(8, 6))
sns.boxplot(data=df, x="sex", y="age", color="#4C72B0")
plt.title("Age Distribution by Sex")
plt.xlabel("Sex")
plt.ylabel("Age")
plt.tight_layout()
plt.savefig("task4_boxplot.png", dpi=100)
plt.show()
```

## Task 5: Outlier Detection — IQR Method

**Why we do this after the box plot:** the box plot in Task 4 showed outliers visually as dots. Now we calculate the exact same thing with numbers, using the formula from the manual — Q1, Q3, and the 1.5×IQR fence. This tells us if what we visually eyeballed in Task 4 matches the actual count.

```python
Q1 = df['age'].quantile(0.25)
Q3 = df['age'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers_iqr = df[(df['age'] < lower) | (df['age'] > upper)]

print(f"Q1 = {Q1}, Q3 = {Q3}, IQR = {IQR}")
print(f"Lower bound = {lower}, Upper bound = {upper}")
print(f"Number of IQR outliers: {len(outliers_iqr)}")
```

## Task 6: Outlier Detection — Z-score Method

**Why compare to Task 5:** IQR and z-score often disagree, and the manual wants us to explain *why* using the shape of the distribution. Since Task 3 showed `age` is right-skewed (not symmetric), z-score is expected to behave differently than IQR here — z-score assumes symmetry around the mean, which this column doesn't really have.

```python
df['zscore'] = stats.zscore(df['age'])
outliers_z = df[df['zscore'].abs() > 3]

print(f"Number of z-score outliers: {len(outliers_z)}")
print(f"Number of IQR outliers:     {len(outliers_iqr)}")
```

**Comparison:** IQR and z-score do *not* flag the same number of outliers. This traces back to Task 3 — `age` is right-skewed (skew = 0.53), not symmetric. IQR only relies on quartiles, so it doesn't care about the shape of the tails and flags any value past the fence regardless of skew. Z-score, on the other hand, assumes the data is roughly bell-shaped and symmetric around the mean; on a skewed column, the right tail pulls the mean and standard deviation upward, which can make z-score more conservative (flag fewer points) on the very high end than a purely quartile-based method would. In short: IQR is the more trustworthy method here since it doesn't assume symmetry.

## Task 7: Scatter Plot — Relationship Between Two Variables

**Why a scatter plot:** we're now asking about the relationship between *two* numeric columns at once — `age` and `hours.per.week`. A scatter plot uses position on both axes simultaneously, which is the strongest possible encoding for this kind of question. Coloring by `income` (categorical) lets us see if the relationship differs between income groups, using color to distinguish categories — exactly what the color channel is good at, per the manual.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x="age", y="hours.per.week", hue="income", alpha=0.3, s=20)
plt.title("Age vs Hours per Week, by Income")
plt.xlabel("Age")
plt.ylabel("Hours per Week")
plt.tight_layout()
plt.savefig("task7_scatter.png", dpi=100)
plt.show()

print("Correlation (age vs hours.per.week):", df['age'].corr(df['hours.per.week']).round(3))
```

## Task 8: Line/Area Chart — Trend Across an Ordered Variable

**Why `education.num`:** the manual specifically calls this out as the ordered variable to use for the Adult dataset, since there's no real time column. `education.num` is a numeric encoding of education level (1 = Preschool ... 16 = Doctorate), so it has a genuine natural order — unlike, say, `occupation`, which has no inherent ranking and shouldn't be connected with a line (that's the "no disconnected bars" caution from the manual, applied in reverse: don't draw a line through unordered categories).

We'll plot average `hours.per.week` across `education.num`, once as a line and once as a filled area, to compare the visual impression.

```python
avg_hours_by_edu = df.groupby("education.num")["hours.per.week"].mean()

fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)

axes[0].plot(avg_hours_by_edu.index, avg_hours_by_edu.values, marker="o", color="#4C72B0")
axes[0].set_title("Line Chart")
axes[0].set_xlabel("Education (years, encoded)")
axes[0].set_ylabel("Average Hours per Week")

axes[1].fill_between(avg_hours_by_edu.index, avg_hours_by_edu.values, color="#4C72B0", alpha=0.5)
axes[1].plot(avg_hours_by_edu.index, avg_hours_by_edu.values, color="#4C72B0")
axes[1].set_title("Area Chart")
axes[1].set_xlabel("Education (years, encoded)")

plt.tight_layout()
plt.savefig("task8_line_area.png", dpi=100)
plt.show()
```

**Comparing the two:** both charts show the exact same trend — average hours worked rises fairly steadily as education level increases. The line chart shows just the direction of change cleanly. The area chart's fill draws your eye to the *magnitude/volume* under the curve, which makes the overall climb look more dramatic even though the underlying numbers are identical. For this data, since we care about the trend direction more than a cumulative "volume," the line chart is the more honest choice — the area chart risks visually overstating how much hours grow with education.

## Task 9: Graphical Integrity Check

**The point of this task:** show how the exact same data can be made to look different just by manipulating the axis — without touching a single number. We'll take the Task 2 count plot and re-plot it with a truncated y-axis (starting above zero instead of at zero).

```python
order = df["education"].value_counts().index
counts = df["education"].value_counts()

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.countplot(data=df, x="education", order=order, color="#4C72B0", edgecolor="black", ax=axes[0])
axes[0].set_title("Honest: Y-axis Starts at 0")
axes[0].set_xlabel("Education")
axes[0].set_ylabel("Count")
axes[0].tick_params(axis='x', rotation=45)

sns.countplot(data=df, x="education", order=order, color="#4C72B0", edgecolor="black", ax=axes[1])
axes[1].set_title("Misleading: Y-axis Truncated")
axes[1].set_xlabel("Education")
axes[1].set_ylabel("Count")
axes[1].tick_params(axis='x', rotation=45)
axes[1].set_ylim(bottom=counts.min() + 50)

plt.tight_layout()
plt.savefig("task9_integrity.png", dpi=100)
plt.show()
```

**Write-up (2-3 sentences):** Even though both charts plot the exact same counts, truncating the y-axis in the second version exaggerates the differences between education levels — bars that were only modestly different in the honest version (like 11th vs 10th vs Prof-school) now look wildly different in height, because the visible range no longer reflects the true zero-based proportions. This is a classic graphical integrity violation: a reader skimming the truncated chart would walk away thinking rare education levels are drastically rarer than common ones, when the real gap, while real, is far less extreme than the chart implies.
