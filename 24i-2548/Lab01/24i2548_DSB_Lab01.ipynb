```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('adult.csv')
```

```python
print("Info:")
df.info()
print("Head:")
display(df.head())
print("Null Sums:")
display(df.isnull().sum())


```

## TASK 1

```python
df.replace('?', np.nan, inplace=True)
display(df.isnull().sum())
```

## TASK 2

```python
missing_count = df.isnull().sum()
missing_percentage = (df.isnull().mean() * 100)

# Combine count and percentage into one table
missing_summary = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing Percentage': missing_percentage
})

missing_summary = missing_summary.sort_values(by='Missing Count', ascending=False)

missing_summary
```

## TASK 3

For "workingclass" and "occuptaion" we chose "unknown" because for categorical data, "unknown" is a reasonable choice because the missing value doesn't necessarily mean that the person belongs to the most common category. Using the mode could incorrectly assign missing people to the most frequent category.

For numerical data we would choose mode if the data was more normal and median if there were larger outliers

```python
df['workclass'] = df['workclass'].fillna('Unknown')
df['occupation'] = df['occupation'].fillna('Unknown')

df[['workclass', 'occupation']].isnull().sum()
```

## TASK 4

```python
exact_duplicates = df.duplicated().sum()

print("Number of exact duplicate rows:", exact_duplicates)
```

```python
duplicate_rows = df[df.duplicated(keep=False)]

duplicate_rows
```

```python
#Remove the target column for feature-level duplicate checking
features_only = df.drop(columns=['income'])

#Count duplicates ignoring the income label
duplicates_without_target = features_only.duplicated().sum()

print(
    "\nDuplicates when ignoring income label:",
    duplicates_without_target
)

#Display feature-level duplicates
duplicate_features = df[features_only.duplicated(keep=False)]

print("\nRows duplicated when ignoring income:")
display(duplicate_features)
```

```python

# Categorical descriptive statistics
print("\nCategorical Descriptive Statistics:")
display(df.describe(include='object'))
```

```python

# Age
age_mean = df['age'].mean()
age_median = df['age'].median()

print("\nAge:")
print("Mean:", age_mean)
print("Median:", age_median)

# Hours per week
hours_mean = df['hours.per.week'].mean()
hours_median = df['hours.per.week'].median()

print("\nHours-per-week:")
print("Mean:", hours_mean)
print("Median:", hours_median)

# Differences
print("\nDifference between mean and median:")
print("Age:", abs(age_mean - age_median))
print("Hours-per-week:", abs(hours_mean - hours_median))
```

The mean age is 38.58 years and the median age is 37 years, giving a difference of approximately 1.58 years. The mean hours-per-week is 40.44 hours while the median is 40 hours, giving a difference of only 0.44 hours. Overall, the mean and median are relatively close for both variables, suggesting that neither distribution differs dramatically between its average and central value.

## TASK 10

### Dataset OverviewDimensions: 
32,561 rows and 15 columns.  Representation: Demographics, employment metadata, and financial metrics extracted from the 1994 US Census dataset to predict individual income class ( <=50K vs >50K ).  

### Data Quality Issues FoundHidden Missing Values:
Initially hidden as '?' strings across categorical fields:  workclass: 1,836 missing values (5.64%)  occupation: 1,843 missing values (5.66%)  native.country: 583 missing values (1.79%)  Duplicates: 24 exact duplicate rows detected across all attributes.  Inconsistent Categorical Data: Strings in categorical columns (education, marital.status, native.country, etc.) contained leading and trailing whitespace.  

### Key ObservationsClass Imbalance:
The target variable income is heavily imbalanced, with 75.92% earning <=50K and 24.08% earning >50K.  Most Common Occupation: Prof-specialty is the single most frequent occupation, representing 12.71% of the dataset.  Feature Redundancy: education and education.num maintain a perfect 1-to-1 deterministic relationship. 

### Cleaning Decisions Made:
Whitespace Trimming: Applied .str.strip() to eliminate leading/trailing whitespace across all text columns.  Missing Value Standardization: Replaced hidden '?' markers with standard np.nan values.  Missing Value Handling: Mode/Unknown category imputation is recommended for workclass, occupation, and native.country to preserve sample size.  Duplicate Removal: Removed 24 exact duplicate records to prevent skewed frequency weights.
