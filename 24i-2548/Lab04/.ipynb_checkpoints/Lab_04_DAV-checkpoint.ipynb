Dataset: food_delivery_orders.csv — 4,200 transaction-level food-delivery orders.
Columns: order_date, city, restaurant_category, cuisine, order_value,
discount_percent, delivery_time_min, rating, profit_per_order.

Every technique follows the Lab Manual literally: histogram+KDE, violin+strip,

## 0. Imports & Data Loading

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
plt.rcParams['figure.dpi'] = 100

df = pd.read_csv('food_delivery_orders.csv')
df['order_date'] = pd.to_datetime(df['order_date'])
print(f'Shape: {df.shape}')
df.head()
```

```python
df.info()
```

### Task 1 — Univariate, properly read

Management wants a single number for "typical delivery time." Plot a histogram
with KDE overlay of delivery_time_min, report its skewness, and compare mean vs
median. In 1–2 sentences, recommend mean or median and justify.

```python
dtm = df['delivery_time_min']
skew = dtm.skew()
mean_val = dtm.mean()
median_val = dtm.median()

print(f'Skewness : {skew:.3f}')
print(f'Mean     : {mean_val:.2f} min')
print(f'Median   : {median_val:.2f} min')
print(f'Std      : {dtm.std():.2f} min')
print(f'Min      : {dtm.min():.2f}  Max: {dtm.max():.2f}')

fig, ax = plt.subplots(figsize=(9, 4.5))
sns.histplot(dtm, bins=40, kde=True, ax=ax, color='steelblue', edgecolor='white')
ax.axvline(mean_val, color='red', ls='--', label=f'Mean = {mean_val:.1f}')
ax.axvline(median_val, color='green', ls='--', label=f'Median = {median_val:.1f}')
ax.set_xlabel('Delivery Time (min)')
ax.set_title(f'Histogram + KDE of delivery_time_min  (skew = {skew:.3f})')
ax.legend()
plt.tight_layout()
plt.show()
```

**Reading:** The distribution is **strongly right-skewed** (skewness = 2.877, well above +1).
The mean (34.5 min) sits above the median (32.5 min) — the classic signature of a long
right tail where a few extremely slow deliveries (up to 132.7 min) pull the average up.

**Recommendation:** Advertise the **median (32.5 min)**. The median is robust to the
slow-tail outliers and represents the experience a *typical* customer actually receives.
The mean would over-promise speed — most customers wait less than 35 min, but the tail
inflates the advertised average.

### Task 2 — Categorical × numerical, beyond the average

For rating by city, build a violin plot and a strip plot (jittered) of the same data.
Which city shows the most inconsistent customer experience?

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)

sns.violinplot(data=df, x='city', y='rating', ax=axes[0],
               inner='quartile', color='lightcoral', cut=0)
axes[0].set_title('Violin plot — rating by city')

sns.stripplot(data=df, x='city', y='rating', ax=axes[1],
              jitter=True, alpha=0.3, size=2, color='navy')
axes[1].set_title('Strip plot (jittered) — rating by city')

for ax in axes:
    ax.set_xlabel('City')
    ax.set_ylabel('Rating')
plt.tight_layout()
plt.show()

city_stats = df.groupby('city')['rating'].agg(['mean', 'median', 'std', 'min', 'max', 'count'])
print(city_stats.round(3))
```

**Answer:** **Multan** shows the most inconsistent customer experience — it has the widest
within-group spread (std = **0.401**) while its mean rating (4.521) is near the overall
average (~4.51). A bar-of-means plot would miss this entirely: every city averages ~4.5,
but the violin and strip plots reveal that Multan has the broadest min-to-max range, meaning
its customers are polarised between very satisfied and quite dissatisfied — the
distributional shape a single number can't convey.

### Task 3 — Multivariate encoding (scatter with hue + size)

Build a single scatter plot of delivery_time_min vs rating with a 3rd variable via
hue (city) and a 4th via size (order_value). State one pattern only visible because
of the 4-variable encoding.

```python
fig, ax = plt.subplots(figsize=(10, 6))
sns.scatterplot(data=df, x='rating', y='delivery_time_min',
                hue='city', size='order_value', sizes=(10, 200),
                alpha=0.6, ax=ax)
ax.set_title('delivery_time_min vs rating  (hue = city, size = order_value)')
ax.set_xlabel('Rating')
ax.set_ylabel('Delivery Time (min)')
ax.legend(bbox_to_anchor=(1.02, 1), loc='upper left', fontsize=8)
plt.tight_layout()
plt.show()
```

```python
print(f'Overall corr(delivery_time_min, rating): {df["delivery_time_min"].corr(df["rating"]):.3f}')
```

**Pattern only visible via 4-variable encoding:** Although the overall correlation
between delivery time and rating is negative (-0.445), the hue reveals that **Lahore**
concentrates the largest markers (highest order_value) in the upper-right cluster.
In other words, many high-value orders in Lahore still receive decent ratings despite
slower delivery, while other cities' large orders cluster at lower delivery times. The
size dimension (order_value) and the hue dimension (city) together separate what looks
like one uniform negative trend into city-specific sub-patterns: high-value orders are
not uniformly rated lower, because the relationship is conditional on city and order
magnitude. This interaction is invisible in any plain bivariate scatter.

### Task 4 — Correlation: Pearson vs Spearman

Compute both correlation matrices for the numeric columns. Identify the pair where the
two methods disagree the most, and determine whether the disagreement is driven by
extreme orders or a genuinely non-linear trend. Decide whether the extreme orders
should be dropped, and justify. Include the correlation≠causation caveat.

```python
num_cols = ['order_value', 'discount_percent', 'delivery_time_min', 'rating', 'profit_per_order']
pearson_m = df[num_cols].corr(method='pearson')
spearman_m = df[num_cols].corr(method='spearman')
diff_m = (pearson_m - spearman_m).abs()

print('=== Pearson ===')
print(pearson_m.round(4))
print()
print('=== Spearman ===')
print(spearman_m.round(4))
print()
print('=== |Pearson - Spearman| ===')
print(diff_m.round(4))
```

```python
# find the pair with the largest disagreement
max_diff = 0
max_pair = None
for i in range(len(num_cols)):
    for j in range(i + 1, len(num_cols)):
        d = abs(pearson_m.iloc[i, j] - spearman_m.iloc[i, j])
        if d > max_diff:
            max_diff = d
            max_pair = (num_cols[i], num_cols[j])

print(f'Largest disagreement: {max_pair[0]} vs {max_pair[1]}')
print(f'  |diff|   = {max_diff:.4f}')
print(f'  Pearson  = {pearson_m.loc[max_pair[0], max_pair[1]]:.4f}')
print(f'  Spearman = {spearman_m.loc[max_pair[0], max_pair[1]]:.4f}')
```

```python
# are extreme orders driving the disagreement?
ov_z = np.abs((df['order_value'] - df['order_value'].mean()) / df['order_value'].std())
print(f'Orders beyond 3 sigma in order_value: {(ov_z > 3).sum()}')
print(f'99th pct of order_value: {df["order_value"].quantile(0.99):.2f}')
print()

df['ov_bin'] = pd.cut(df['order_value'], bins=10)
binned = df.groupby('ov_bin', observed=True)['profit_per_order'].agg(['mean', 'median', 'count'])
print(binned.round(1))
```

```python
# re-check after capping extremes
ov_cap = df['order_value'].quantile(0.99)
df_cap = df[df['order_value'] <= ov_cap]
print(f'Before capping: n={len(df)}')
print(f'After  capping (<= {ov_cap:.0f}): n={len(df_cap)}')
print(f'  Pearson  after = {df_cap[max_pair[0]].corr(df_cap[max_pair[1]], method="pearson"):.4f}')
print(f'  Spearman after = {df_cap[max_pair[0]].corr(df_cap[max_pair[1]], method="spearman"):.4f}')
```

**Largest disagreement:** `order_value` vs `profit_per_order`  |Pearson − Spearman| = 1.3303

  - Pearson  = -0.6571  → negative (higher order_value → lower profit)
  - Spearman = 0.6731  → positive (rank-order is monotonic increasing)

The disagreement is **driven by a handful of extreme orders**. The binned table shows
that for order values up to ~9,500, binned mean profit increases steadily (a positive
monotonic trend → Spearman ~ +0.67). But the ~70 orders above the 99th percentile
(9924) are mega-orders that all carry heavy losses (profit ~ -4,000), dragging
Pearson down to -0.66. At 99th-percentile capping, Pearson becomes 0.10 and
Spearman 0.72 — both positive and aligned, confirming the core relationship is
positive and monotonic.

**Should extremes be dropped?** **Yes**, for the headline figure. Dropping the 42 orders
above the 99th percentile leaves 4158 rows. The remaining Pearson (0.10) and Spearman
(0.72) are close, so the *true* business relationship is positive. Report Spearman
(0.67) as the robust headline, and flag the loss-making mega-orders separately.

> **Correlation is not causation:** a correlation between order_value and profit does
**not** mean that raising order value *causes* higher profit. Both are jointly driven by
confounders — restaurant category, cuisine, and discount strategy. Management must not
treat the correlation as a lever they can pull directly.

### Task 5 — Layered aggregation: pivot_table + heatmap

Build a city × restaurant_category table of average profit_per_order with margins=True.
Render as a heatmap with a diverging palette. State the most profitable combination,
and note whether it is also the top revenue generator.

```python
pt = pd.pivot_table(df, index='city', columns='restaurant_category',
                    values='profit_per_order', aggfunc='mean', margins=True)
print(pt.round(1))
```

```python
pt_no = pt.drop(index='All', columns='All')
best_idx = pt_no.stack().idxmax()
best_val = pt_no.loc[best_idx[0], best_idx[1]]
print(f'Most profitable: {best_idx[0]} / {best_idx[1]} = {best_val:.2f} avg profit/order')

pt_rev = pd.pivot_table(df, index='city', columns='restaurant_category',
                        values='order_value', aggfunc='sum', margins=True)
pt_rev_no = pt_rev.drop(index='All', columns='All')
rev_idx = pt_rev_no.stack().idxmax()
rev_val = pt_rev_no.loc[rev_idx[0], rev_idx[1]]
print(f'Top revenue: {rev_idx[0]} / {rev_idx[1]} = {rev_val:.0f} total revenue')
```

```python
pt_no = pt.drop(index='All', columns='All')
fig, ax = plt.subplots(figsize=(8, 5))
sns.heatmap(pt_no, annot=True, fmt='.1f', cmap='RdBu_r',
            center=0, ax=ax, cbar_kws={'label': 'Avg profit_per_order'})
ax.set_title('Avg profit_per_order by city x restaurant_category')
plt.tight_layout()
plt.show()
```

**Most profitable combination:** `Karachi` / `Fine Dining` with an average of **302.8** profit per
order — the brightest cell in the heatmap.

**Revenue vs profitability:** That most-profitable combination is **not** the top revenue
generator — the highest-revenue cell is `Lahore` / `Cloud Kitchen` (total revenue ~ 449767). High
revenue and high profitability point to **different** city-category combinations, so
management should track both metrics separately rather than assuming the busiest segment
is also the most profitable.

### Task 6 — Time-series: trend vs noise

Resample to get average monthly rating and overlay a 3-month rolling average.
State whether satisfaction trends up, down, or flat once noise is smoothed, and
whether the raw monthly line alone would have led to a different conclusion.

```python
df_ts = df.set_index('order_date')
monthly = df_ts['rating'].resample('ME').mean()
rolling = monthly.rolling(window=3).mean()

fig, ax = plt.subplots(figsize=(11, 5))
monthly.plot(ax=ax, label='Monthly avg rating', alpha=0.5, marker='o', ms=3, color='steelblue')
rolling.plot(ax=ax, label='3-month rolling avg', linewidth=2, color='crimson')
ax.set_ylabel('Average Rating')
ax.set_title('Monthly average rating vs 3-month rolling average')
ax.legend()
plt.tight_layout()
plt.show()
```

**Trend (smoothed):** Once the 3-month rolling average is applied, customer
satisfaction is **essentially flat with a slight decline**. The rolling line starts
near 4.526 (Jan 2023) and ends near 4.513 (Jun 2024) — a net change of -0.013 over ~18 months.

**Would the raw line mislead?** Yes. The raw monthly average swings only between
4.451 and 4.545 (a range of ~0.09), and month-to-month jumps of 0.03 can look
meaningful in isolation. The rolling average reveals that these fluctuations are
**noise**, not a signal — the true underlying trend is flat, and an observer looking
only at the raw line might mistake a single uptick or dip for a real shift.

### Task 7 — Faceting: small multiples

Using FacetGrid, facet delivery_time_min vs rating by restaurant_category (col) and
by whether discount_percent > 20 (hue). State one category where heavy discounting
visibly changes how forgiving customers are of slow delivery.

```python
# create heavy_discount column for this task
df['heavy_discount'] = df['discount_percent'] > 20
print(df.groupby('heavy_discount').size())
```

```python
g = sns.FacetGrid(df, col='restaurant_category', hue='heavy_discount',
                  col_wrap=3, height=4, aspect=1.1, sharex=False, sharey=False)
g.map_dataframe(sns.scatterplot, x='rating', y='delivery_time_min',
                alpha=0.5, s=15, edgecolor='none')
g.add_legend(title='Heavy discount (>20%)')
g.set_titles('{col_name}')
g.set_axis_labels('Rating', 'Delivery Time (min)')
g.fig.suptitle('delivery_time_min vs rating by category and discount level',
                y=1.03, fontsize=12)
plt.show()
```

```python
# quantify: delivery_time-rating correlation, heavy vs light discount, per category
for cat in df['restaurant_category'].unique():
    sub = df[df['restaurant_category'] == cat]
    h = sub[sub['heavy_discount']]
    l = sub[~sub['heavy_discount']]
    if len(h) > 5 and len(l) > 5:
        rh = h['delivery_time_min'].corr(h['rating'])
        rl = l['delivery_time_min'].corr(l['rating'])
        print(f'{cat:15s}  heavy n={len(h):4d} r={rh:+.3f} |'
              f' light n={len(l):4d} r={rl:+.3f} | gap={rh - rl:+.3f}')
```

**Category where discounting changes forgiveness:** **Fine Dining**. The largest gap between
the delivery-time–rating correlation under heavy vs light discounting appears here:
with heavy discounts the correlation is r = -0.737 (n=70), versus r = -0.402 for light
discounts (n=362). In other words, heavy discounting shifts customers' tolerance — they
become more forgiving of a slow delivery in this category, visibly flattening the
negative trend in the facet panel.

### Task 8 — Overplotting + annotation (combined)

Plot the raw delivery_time_min vs rating scatter (expect heavy overplotting from
repeated rating values). Fix it, justify the fix, and annotate the order with the
**best rating despite an unusually long delivery** and the order with the **worst
rating despite an unusually fast delivery**.

```python
# diagnose overplotting
print(f'Total orders  : {len(df)}')
print(f'Unique ratings: {df["rating"].nunique()}  (ratings binned to 1 decimal -> heavy overlap)')
```

```python
# raw scatter — overplotted
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(df['rating'], df['delivery_time_min'], alpha=0.2, s=8, color='navy')
ax.set_xlabel('Rating')
ax.set_ylabel('Delivery Time (min)')
ax.set_title('Raw scatter — heavy overplotting')
plt.tight_layout()
plt.show()
```

```python
# fix with alpha transparency and annotate extremes
long_del = df[df['delivery_time_min'] >= df['delivery_time_min'].quantile(0.90)]
best_long = long_del.loc[long_del['rating'].idxmax()]
fast_del = df[df['delivery_time_min'] <= df['delivery_time_min'].quantile(0.10)]
worst_fast = fast_del.loc[fast_del['rating'].idxmin()]

fig, ax = plt.subplots(figsize=(9, 6))
ax.scatter(df['rating'], df['delivery_time_min'],
             alpha=0.3, s=12, color='steelblue', edgecolor='none', label='all orders')

ax.scatter([best_long['rating']], [best_long['delivery_time_min']],
             s=120, color='green', marker='^', zorder=5, edgecolor='black', linewidth=0.5)
ax.annotate(f"Best rating, slow delivery\n({best_long['rating']:.1f}, {best_long['delivery_time_min']:.1f} min)",
            xy=(best_long['rating'], best_long['delivery_time_min']),
            xytext=(best_long['rating'] + 0.15, best_long['delivery_time_min'] - 15),
            arrowprops=dict(arrowstyle='->', color='green', lw=1.5),
            fontsize=9, color='green')

ax.scatter([worst_fast['rating']], [worst_fast['delivery_time_min']],
             s=120, color='red', marker='v', zorder=5, edgecolor='black', linewidth=0.5)
ax.annotate(f"Worst rating, fast delivery\n({worst_fast['rating']:.1f}, {worst_fast['delivery_time_min']:.1f} min)",
            xy=(worst_fast['rating'], worst_fast['delivery_time_min']),
            xytext=(worst_fast['rating'] - 0.3, worst_fast['delivery_time_min'] + 8),
            arrowprops=dict(arrowstyle='->', color='red', lw=1.5),
            fontsize=9, color='red')

ax.set_xlabel('Rating')
ax.set_ylabel('Delivery Time (min)')
ax.set_title('Fixed scatter (alpha=0.3) with annotated extremes')
plt.tight_layout()
plt.show()
print(f'Best rating despite long delivery:  rating={best_long["rating"]:.1f}, '
      f'time={best_long["delivery_time_min"]:.1f} min, city={best_long["city"]}')
print(f'Worst rating despite fast delivery: rating={worst_fast["rating"]:.1f}, '
      f'time={worst_fast["delivery_time_min"]:.1f} min, city={worst_fast["city"]}')
```

**Why alpha over hexbin:** With 4200 points and only 27 discrete rating levels, a raw
scatter collapses into a near-solid block. **Alpha transparency** (alpha = 0.3) is the
right fix because the x-axis (rating) is discrete — hexbin or KDE would erase the exact
rating values we need for annotation, whereas alpha preserves every point and reveals
density through overlap.

**Annotated insights:**
- Green up-triangle: the **best rating despite an unusually long delivery** — rating
  5.0 at 79.5 min (top-decile slowest). A slow delivery alone does not doom a rating.
- Red down-triangle: the **worst rating despite an unusually fast delivery** — rating
  3.9 at 21.9 min (top-decile fastest). A fast delivery cannot always compensate for
  other issues (food quality, order accuracy, cold food, etc.).
